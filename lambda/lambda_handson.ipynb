#### Reserved & Provisioned Concurrency

* **Reserved Concurrency**: Limits the maximum number of concurrent executions for a function.
* **Provisioned Concurrency**: Pre-warms a number of Lambda instances to handle requests with **zero cold start**.

- ⚡️ **Set Reserved Concurrency**:

    ```python
    def set_reserved_concurrency(function_name, reserved_concurrent_executions):
        response = lambda_client.put_function_concurrency(
            FunctionName=function_name,
            ReservedConcurrentExecutions=reserved_concurrent_executions
        )
        return response
    ```

    -   **Remove Reserved Concurrency**:

        ```python
        def delete_reserved_concurrency(function_name):
            response = lambda_client.delete_function_concurrency(
                FunctionName=function_name
            )
            return response
        ```

- ⚡️ **Set Provisioned Concurrency**: To use provisioned concurrency, you must:

    🔹 Step 1: **Publish a version** of the Lambda function.

        ```python
        def publish_new_version(function_name):
            response = lambda_client.publish_version(FunctionName=function_name)
            return response['Version']
        ```

    🔹 Step 2: Configure concurrency on that version (not `$LATEST`).

        ```python
        def set_provisioned_concurrency(function_name, version, provisioned_count):
            response = lambda_client.put_provisioned_concurrency_config(
                FunctionName=function_name,
                Qualifier=version,  # e.g. '1'
                ProvisionedConcurrentExecutions=provisioned_count
            )
            return response
        ```

    - **Example Usage**

        ```python
        fn_name = "my-lambda-fn"

        # Set reserved concurrency
        set_reserved_concurrency(fn_name, 10)

        # Publish version
        version = publish_new_version(fn_name)

        # Set provisioned concurrency for that version
        set_provisioned_concurrency(fn_name, version, 5)
        ```


    - **Delete provisioned concurrency**:

        ```python
        lambda_client.delete_provisioned_concurrency_config(
            FunctionName=fn_name,
            Qualifier=version
        )
        ```

    - **Notes**:
        * **Provisioned concurrency costs extra** even if your function isn't called.
        * Monitor both types of concurrency via **CloudWatch metrics**.
        * For aliases (e.g. "prod"), use the alias name as `Qualifier`.


#### Autoscaling Lambda

Auto Scaling for **AWS Lambda** is done by dynamically adjusting **Provisioned Concurrency** based on usage metrics. While Lambda automatically scales with demand, configuring **auto scaling of provisioned concurrency** ensures low latency (no cold starts) while managing costs. Here’s a complete breakdown of how to **configure Lambda autoscaling** using **Application Auto Scaling**.

-   **When Do You Need Auto Scaling for Lambda?**

    * If you’re using **provisioned concurrency**.
    * Want to **avoid cold starts** for high-throughput APIs.
    * Need **predictable latency** for critical applications.


##### 🛠 Step-by-Step: Setup Lambda Auto Scaling

* `lambda` – to publish version and configure concurrency
* `application-autoscaling` – to register the Lambda as a scalable target


1. **Prerequisites**:


    ```python
    import boto3

    lambda_client = boto3.client('lambda', region_name='us-east-1')
    autoscaling_client = boto3.client('application-autoscaling', region_name='us-east-1')
    ```

2. **Publish a Version of the Lambda Function**: Provisioned concurrency must be attached to a **version** (not `$LATEST`).

    ```python
    response = lambda_client.publish_version(
        FunctionName='my-lambda-fn'
    )
    version = response['Version']
    ```

3. **Register Scalable Target**:

    ```python
    autoscaling_client.register_scalable_target(
        ServiceNamespace='lambda',
        ResourceId=f'function:my-lambda-fn:{version}',
        ScalableDimension='lambda:function:ProvisionedConcurrency',
        MinCapacity=1,
        MaxCapacity=20
    )
    ```

4. **Create Scaling Policy**: Auto Scaling supports **Target Tracking Scaling Policies** (similar to EC2 ASG).

    ```python
    autoscaling_client.put_scaling_policy(
        PolicyName='lambda-provisioned-concurrency-scaling',
        ServiceNamespace='lambda',
        ResourceId=f'function:my-lambda-fn:{version}',
        ScalableDimension='lambda:function:ProvisionedConcurrency',
        PolicyType='TargetTrackingScaling',
        TargetTrackingScalingPolicyConfiguration={
            'TargetValue': 0.75,  # utilization %
            'PredefinedMetricSpecification': {
                'PredefinedMetricType': 'LambdaProvisionedConcurrencyUtilization'
            },
            'ScaleInCooldown': 60,
            'ScaleOutCooldown': 60
        }
    )
    ```

-   **Key Concepts**:

    | Term                        | Description                                               |
    | --------------------------- | --------------------------------------------------------- |
    | **Provisioned Concurrency** | Pre-warms Lambda instances; avoids cold starts.           |
    | **Scalable Target**         | Lambda version (or alias) registered to auto scale.       |
    | **Target Tracking Policy**  | Maintains a utilization target (e.g., 75%).               |
    | **Metric**                  | `LambdaProvisionedConcurrencyUtilization` measures usage. |


-   **Optional: Use an Alias Instead of Version**:

    Aliases allow you to manage versions more cleanly:

    ```python
    lambda_client.create_alias(
        FunctionName='my-lambda-fn',
        Name='prod',
        FunctionVersion='2',
        Description='Production alias'
    )
    ```

    Then use:

    ```python
    ResourceId = 'function:my-lambda-fn:prod'
    ```

-   **Cleanup**

    To delete the scalable target:

    ```python
    autoscaling_client.deregister_scalable_target(
        ServiceNamespace='lambda',
        ResourceId=f'function:my-lambda-fn:{version}',
        ScalableDimension='lambda:function:ProvisionedConcurrency'
    )
    ```

- **Notes**

    * **Auto scaling only applies to provisioned concurrency**, not regular on-demand Lambda scaling.
    * You can monitor metrics in **CloudWatch > Lambda Insights**.
    * Costs increase with provisioned concurrency. Use judiciously.


#### Configure EFS of Lambda

Amazon EFS (Elastic File System) provides a scalable, fully managed, and highly available file storage service that can be integrated with AWS Lambda. This combination is particularly useful for Lambda functions that need to read or write large amounts of data, share files between invocations, or work with data that's too big to fit in the `/tmp` directory. Following are the usecases of EFS with Lambda.

* **Handling Large Datasets**: When a Lambda function needs to process or analyze a large file (e.g., a video file, a large CSV, or a machine learning model) that exceeds the 512 MB `/tmp` directory limit, EFS offers a perfect solution. The function can access the file directly from the EFS volume without having to download it, which saves time and avoids memory constraints.

* **Shared Data and Persistent State**: Lambda is stateless, meaning each invocation is an independent event. EFS allows multiple Lambda functions or multiple invocations of the same function to access and share the same data. This is useful for:
    * **Content Management**: Storing and serving static content like images, documents, or web assets that multiple functions need to access.
    * **Caching**: Storing frequently accessed data or pre-computed results in a shared location to avoid re-computation.
    * **Data Consistency**: Sharing a central repository for application state or configuration files that need to be consistent across all function invocations.

* **Machine Learning (ML)**: ML models and their dependencies can be very large. Storing them on an EFS volume allows a Lambda function to load a pre-trained model quickly for inference without having to bundle it with the deployment package, which has a size limit of 250 MB.

* **Centralized Logging and File Archiving**: When you need a centralized location to store logs or other output files generated by multiple Lambda functions, EFS is a great option. It provides a persistent and scalable file system for long-term storage and easy access.

* **Building and Deployment**: EFS can be used as a shared drive for build processes. For example, multiple functions could access a common library or a set of tools to compile and package code, streamlining the CI/CD pipeline.

To configure **Amazon EFS (Elastic File System)** with **AWS Lambda** using `boto3` in Python, you must:

1. Create the EFS file system and mount targets.
2. Create an EFS access point.
3. Configure the Lambda function to use the access point.
4. Ensure that the Lambda function is in the same VPC and subnets as the EFS mount targets.

-   **Prerequisites**:

    * Lambda function is in a **VPC** with access to EFS mount targets.
    * VPC security group allows NFS traffic (`port 2049`) from Lambda to EFS.
    * IAM Role for Lambda has the required EFS access permissions.


##### 🛠 Step-by-Step configuration of EFS with Lambda

-   Role: `AWSLambdaVPCAccessExecutionRole`


1. **Create EFS File System**

    ```python
    import boto3

    efs = boto3.client('efs', region_name='us-east-1')

    response = efs.create_file_system(
        CreationToken='lambda-efs-demo',
        PerformanceMode='generalPurpose',
        Encrypted=True,
        Tags=[{'Key': 'Name', 'Value': 'lambda-efs-fs'}]
    )
    file_system_id = response['FileSystemId']
    print("EFS FileSystem ID:", file_system_id)
    ```

2. **Create Mount Targets for EFS (one in each subnet)**

    ```python
    subnet_ids = ['subnet-abc123', 'subnet-def456']
    security_group_id = 'sg-0123456789abcdef0'

    for subnet_id in subnet_ids:
        efs.create_mount_target(
            FileSystemId=file_system_id,
            SubnetId=subnet_id,
            SecurityGroups=[security_group_id]
        )
    ```

3. **Create an Access Point**

    ```python
    access_point = efs.create_access_point(
        FileSystemId=file_system_id,
        PosixUser={
            'Uid': 1000,
            'Gid': 1000
        },
        RootDirectory={
            'Path': '/lambda',
            'CreationInfo': {
                'OwnerUid': 1000,
                'OwnerGid': 1000,
                'Permissions': '750'
            }
        },
        Tags=[{'Key': 'Name', 'Value': 'lambda-efs-ap'}]
    )

    access_point_arn = access_point['AccessPointArn']
    print("Access Point ARN:", access_point_arn)
    ```

4. **Attach EFS Access Point to Lambda**

    ```python
    lambda_client = boto3.client('lambda', region_name='us-east-1')

    response = lambda_client.update_function_configuration(
        FunctionName='my-lambda-function',
        FileSystemConfigs=[
            {
                'Arn': access_point_arn,
                'LocalMountPath': '/mnt/efs'
            }
        ]
    )
    ```

5. **Update Lambda Execution Role (If Needed)**

    Attach IAM permissions to Lambda execution role:

    ```json
    {
    "Effect": "Allow",
    "Action": [
        "elasticfilesystem:ClientMount",
        "elasticfilesystem:ClientWrite",
        "elasticfilesystem:ClientRootAccess"
    ],
    "Resource": "<EFS-AccessPoint-ARN>"
    }
    ```

    Also add:

    ```json
    {
    "Effect": "Allow",
    "Action": "elasticfilesystem:DescribeMountTargets",
    "Resource": "*"
    }
    ```

-   **Inside Lambda Function Code**:

    Now, in your Lambda code, you can access the EFS volume:

    ```python
    with open("/mnt/efs/test.txt", "w") as f:
        f.write("Hello from Lambda to EFS!")
    ```

-   **Verifying the Configuration**:

    * Go to **AWS Lambda Console > Your function > Configuration > File Systems** – ensure your mount is listed.
    * Check that the Lambda function is in the **same VPC** as EFS.
    * Monitor errors in **CloudWatch Logs** for permission issues or mount failures.


#### Error Handling, Retry Policies, and Destinations of AWS Lambda


-   **Prerequisites**


    * A Lambda function already created.
    * Valid destination ARNs for success or failure cases (e.g., SNS, SQS, or another Lambda).
    * IAM permissions for `lambda:UpdateFunctionConfiguration` and `lambda:PutFunctionEventInvokeConfig`.

1. **Error Handling and Retry Policies (for Async Invocations)**:

    * `MaximumRetryAttempts`: How many times Lambda retries async invocations (0–2).
    * `MaximumEventAgeInSeconds`: Max time (60 – 21600 sec) Lambda retries events before discarding them.


    ```python
    import boto3

    lambda_client = boto3.client('lambda', region_name='us-east-1')

    response = lambda_client.put_function_event_invoke_config(
        FunctionName='your-lambda-function-name',
        MaximumRetryAttempts=1,
        MaximumEventAgeInSeconds=3600
    )

    print("Async error handling and retry policy configured.")
    ```

2. **Configure Destinations for Async Invocation**:

    -   Lambda supports two types of destinations:

        * `OnSuccess`: Invoked when Lambda executes successfully.
        * `OnFailure`: Invoked after retries are exhausted.

    -   Supported targets: `SNS`, `SQS`, `Lambda`, `EventBridge`


    ```python
    # Example with SQS/SNS destination
    response = lambda_client.put_function_event_invoke_config(
        FunctionName='your-lambda-function-name',
        DestinationConfig={
            'OnSuccess': {
                'Destination': 'arn:aws:sqs:us-east-1:123456789012:success-queue'
            },
            'OnFailure': {
                'Destination': 'arn:aws:sns:us-east-1:123456789012:failure-topic'
            }
        }
    )

    print("Destinations configured for success and failure.")
    ```

3. **Verify Configuration**:

    ```python
    config = lambda_client.get_function_event_invoke_config(
        FunctionName='your-lambda-function-name'
    )

    print("Current config:", config)
    ```

4. **Handling Errors in Code with Try-Except**:

    Within the Lambda handler, you should still write defensive code:

    ```python
    def lambda_handler(event, context):
        try:
            # Your logic here
            print("Processing event:", event)
            return {"status": "success"}
        except Exception as e:
            print(f"Error occurred: {e}")
            raise e  # Ensure errors propagate for retry/destination handling
    ```

-   **Optional: Remove Config**:

    To delete async invoke config:

    ```python

        FunctionName='your-lambda-function-name'
    )
    ```
